```python
import pandas as pd 
import json
```

```python
csv_file = pd.read_csv('/Users/nalex2023/Downloads/references_spreadsheet - Sheet1.csv')

csv_file_removed = csv_file.dropna(axis=0,how='any', subset=['Approximate lat', 'Approximate lon'])

csv_file_removed2 = csv_file_removed.dropna(axis=0,how='any', subset=['Year'] )

csv_file_removed2
```

```python
csv_file_removed2['id'] = range(1, len(csv_file_removed2) + 1)

csv_file_removed2['Title'] = csv_file_removed2['Observation type']

csv_file_removed2['Authors'] = csv_file_removed2['Unnamed: 0']

csv_file_removed2['Year'] = csv_file_removed2['Year'].astype(int,errors='ignore')

csv_file_removed2['Author Keywords'] = csv_file_removed2['Observation type']
csv_file_removed2['Source'] = 'dummy source'

final_df = csv_file_removed2[['id', 'Title', 'Authors', 'Year', 'Author Keywords', 'Source',]]

```

```python
final_df['Title'] = final_df['Title'].str.replace(',', ';')

```

```python
final_df['Title'] = final_df['Title'].fillna('Surface')

final_df['Author Keywords'] = final_df['Author Keywords'].fillna('Surface')

final_df['Author Keywords'] = final_df['Author Keywords'].str.replace(',', ';')


final_df
```

```python
final_df.to_csv('/Users/nalex2023/main/LSB_main/Review_Paper/sample_test_dft.csv', index=False)
```
